```python
import tracemalloc

# Жадібний підхід
tracemalloc.start()
_ = sum([x * x for x in range(10**7)])
_, peak_eager = tracemalloc.get_traced_memory()
tracemalloc.stop()

# Лінивий підхід
tracemalloc.start()
_ = sum(x * x for x in range(10**7))
_, peak_lazy = tracemalloc.get_traced_memory()
tracemalloc.stop()

print(f"Жадібний: {peak_eager / (1024 * 1024):.2f} MB")
print(f"Лінивий:  {peak_lazy / 1024:.2f} KB")
```

```python
def ordinary(n):
    print("start", n)
    return n

def countdown(n):
    print("start", n)
    while n > 0:
        yield n
        n -= 1

print("Виклик ordinary:")
res = ordinary(3)

print("\nВиклик countdown:")
gen = countdown(3)
print(type(gen))

print("\nПерший next(gen):")
print(next(gen))
```

```python
import unicodedata

s1 = "café"
s2 = "cafe\u0301"

print(f"Порівняння на вигляд: '{s1}' vs '{s2}'")
print("Пряме порівняння:", s1 == s2)
print("Довжина s1 і s2:", len(s1), "і", len(s2))

# Нормалізація
norm_s1 = unicodedata.normalize("NFC", s1)
norm_s2 = unicodedata.normalize("NFC", s2)

print("Після NFC:", norm_s1 == norm_s2)
print("Довжина norm_s1 і norm_s2:", len(norm_s1), "і", len(norm_s2))
```

```python
from pathlib import Path

test_file = Path("test_stream.txt")
test_file.write_text("line 1\nline 2\nline 3\n", encoding="utf-8")

with test_file.open("r", encoding="utf-8") as f:
    print("Перший прохід:")
    for line in f:
        print(line.strip())

    print("\nДругий прохід:")
    for line in f:
        print(line.strip())

test_file.unlink()
```
